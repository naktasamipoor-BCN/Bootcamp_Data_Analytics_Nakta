## Ejercicio 1: Calculadora del Índice de Masa Corporal

El objetivo de este ejercicio es crear una función que calcule el Índice de Masa Corporal (IMC) a partir del peso y la altura de una persona.

La fórmula del IMC es:

IMC = peso / altura²

La función también valida los valores de entrada y clasifica el resultado del IMC en su categoría correspondiente.

```python
def calculate_bmi(weight, height):
    """
    Calculate the Body Mass Index (BMI) and classify it into a category.

    Parameters:
    weight (float): Weight in kilograms.
    height (float): Height in meters.

    Returns:
    dict: BMI value and its category, or an error message for invalid inputs.
    """
    
    if not isinstance(weight, (int, float)) or not isinstance(height, (int, float)):
        return {
            "error": "Weight and height must be numeric values."
        }
    
    if weight <= 0 or height <= 0:
        return {
            "error": "Weight and height must be greater than zero."
        }
    
    if weight < 20 or weight > 300:
        return {
            "error": "Weight value is not reasonable."
        }
    
    if height < 0.5 or height > 2.5:
        return {
            "error": "Height value is not reasonable. Height must be provided in meters."
        }
    
    bmi = weight / (height ** 2)
    bmi = round(bmi, 2)
    
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal weight"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obesity"
    
    return {
        "bmi": bmi,
        "category": category
    }
```

## Ejercicio 2: Convertidor de temperaturas

El objetivo de este ejercicio es crear una función que convierta una temperatura desde una unidad de origen a varias unidades de destino.

En este caso, la función acepta como unidades de entrada Celsius, Fahrenheit y Kelvin.  
A partir de la unidad introducida, primero convierte el valor a Celsius y después calcula el resto de conversiones.

El resultado se devuelve como un diccionario para que pueda almacenarse y reutilizarse posteriormente.

```python
def convert_temperature(temperature, from_unit):
    """
    Convert a temperature value from Celsius, Fahrenheit or Kelvin
    into several temperature units.

    Parameters:
    temperature (float): Temperature value to convert.
    from_unit (str): Original temperature unit. Accepted values: 'C', 'F', 'K'.

    Returns:
    dict: Converted temperature values or an error message for invalid inputs.
    """
    
    if not isinstance(temperature, (int, float)):
        return {
            "error": "Temperature must be a numeric value."
        }
    
    if not isinstance(from_unit, str):
        return {
            "error": "Temperature unit must be a string."
        }
    
    from_unit = from_unit.upper()
    
    if from_unit not in ["C", "F", "K"]:
        return {
            "error": "Invalid unit. Use 'C' for Celsius, 'F' for Fahrenheit or 'K' for Kelvin."
        }
    
    if from_unit == "C":
        celsius = temperature
    elif from_unit == "F":
        celsius = (temperature - 32) * 5 / 9
    else:
        celsius = temperature - 273.15
    
    if celsius < -273.15:
        return {
            "error": "Temperature cannot be below absolute zero."
        }
    
    conversions = {
        "Celsius": round(celsius, 2),
        "Fahrenheit": round((celsius * 9 / 5) + 32, 2),
        "Kelvin": round(celsius + 273.15, 2),
        "Rankine": round((celsius + 273.15) * 9 / 5, 2),
        "Reaumur": round(celsius * 4 / 5, 2)
    }
    
    return conversions
```

## Ejercicio 3: Contador de palabras de un texto

El objetivo de este ejercicio es crear una función que reciba un texto y cuente cuántas veces aparece cada palabra.

Antes de contar las palabras, el texto se transforma a minúsculas y se eliminan los signos de puntuación para evitar que palabras como `Hola`, `hola,` y `hola` se cuenten como palabras diferentes.

Como funcionalidad extra, también se calcula la longitud media de las palabras del texto.

```python
def count_words(text):
    """
    Count the frequency of each word in a text and calculate the average word length.

    Parameters:
    text (str): Input text.

    Returns:
    dict: Word frequencies and average word length, or an error message for invalid inputs.
    """
    
    if not isinstance(text, str):
        return {
            "error": "Input must be a text string."
        }
    
    if text.strip() == "":
        return {
            "error": "Input text cannot be empty."
        }
    
    text = text.lower()
    cleaned_text = re.sub(r"[^\w\sáéíóúàèìòùäëïöüñç]", "", text)
    words = cleaned_text.split()
    
    if len(words) == 0:
        return {
            "error": "No valid words found in the text."
        }
    
    word_frequencies = dict(Counter(words))
    average_word_length = sum(len(word) for word in words) / len(words)
    
    return {
        "word_frequencies": word_frequencies,
        "average_word_length": round(average_word_length, 2)
    }
```

## Ejercicio 4: Diccionario inverso con posibilidad de duplicados

El objetivo de este ejercicio es crear una función que invierta un diccionario, intercambiando sus claves y valores.

Si todos los valores del diccionario original son únicos, la inversión es directa.  
Sin embargo, si existen valores duplicados, la función debe evitar la pérdida de información agrupando las claves originales asociadas a ese valor en una lista.

Además, la función devuelve un mensaje de advertencia cuando detecta valores duplicados.

```python
def reverse_dictionary(input_dict):
    """
    Reverse the keys and values of a dictionary.
    If duplicated values are found, all related keys are stored in a list.

    Parameters:
    input_dict (dict): Dictionary to reverse.

    Returns:
    dict: Reversed dictionary and warning information if duplicated values exist.
    """
    
    if not isinstance(input_dict, dict):
        return {
            "error": "Input must be a dictionary."
        }
    
    if len(input_dict) == 0:
        return {
            "error": "Input dictionary cannot be empty."
        }
    
    reversed_dict = {}
    duplicated_values = {}
    
    for key, value in input_dict.items():
        if value in reversed_dict:
            if not isinstance(reversed_dict[value], list):
                reversed_dict[value] = [reversed_dict[value]]
            
            reversed_dict[value].append(key)
            duplicated_values[value] = reversed_dict[value]
        else:
            reversed_dict[value] = key
    
    if duplicated_values:
        return {
            "warning": "Duplicated values were found. Related keys have been grouped in lists.",
            "duplicated_values": duplicated_values,
            "reversed_dictionary": reversed_dict
        }
    
    return {
        "reversed_dictionary": reversed_dict
    }
```

### Ejecución interactiva

Además de probar la función con diccionarios predefinidos, también se puede crear una versión interactiva que permita al usuario construir un diccionario manualmente.

En lugar de pedir al usuario que escriba todo el diccionario en una sola línea, la función solicita pares de clave y valor uno por uno.  
Cuando el usuario escribe `stop`, la entrada de datos termina y se llama a la función principal `reverse_dictionary()`.

De esta forma, no es necesario utilizar librerías adicionales y la función principal se mantiene separada de la entrada de datos.

```python
def run_reverse_dictionary():
    """
    Build a dictionary interactively by asking the user for key-value pairs,
    then reverse it using the reverse_dictionary function.

    Returns:
    dict: Reversed dictionary, warning information, or an error message.
    """
    
    user_dict = {}
    
    while True:
        key = input("Enter a key or type 'stop' to finish: ")
        
        if key.lower() == "stop":
            break
        
        value = input("Enter the value for this key: ")
        user_dict[key] = value
    
    return reverse_dictionary(user_dict)
```

## Ejercicio 1: Contador y organizador de palabras desde un archivo TXT

El objetivo de este ejercicio es leer un archivo `.txt`, contar la frecuencia de cada palabra y organizar el resultado según la letra inicial de cada palabra.

La salida será un diccionario anidado.  
La primera clave representa la letra inicial y, dentro de cada letra, se almacenan las palabras ordenadas alfabéticamente junto con su frecuencia.

Por ejemplo, la estructura esperada es:

{
    "a": {
        "agua": 1,
        "alba": 4
    },
    "b": {
        "baco": 1,
        "blanca": 3
    }
}

El archivo `.txt` se encuentra en la misma carpeta que el notebook, por lo que se utiliza una ruta relativa. Esto permite que el proyecto sea más reproducible al subirlo a GitHub.

```python
def count_and_sort_words_from_file(file_path):
    """
    Read a TXT file, count word frequencies and organize words by their first letter.

    Parameters:
    file_path (str): Path to the TXT file.

    Returns:
    dict: A nested dictionary where each key is a first letter and each value is
          another dictionary with words and their frequencies.
    """
    
    if not isinstance(file_path, str):
        return {
            "error": "File path must be a string."
        }
    
    try:
        with open(file_path, "r", encoding="utf-8") as file:
            text = file.read()
    except FileNotFoundError:
        return {
            "error": "File not found. Please check the file path."
        }
    except Exception as error:
        return {
            "error": f"An unexpected error occurred: {error}"
        }
    
    if text.strip() == "":
        return {
            "error": "The file is empty."
        }
    
    text = text.lower()
    cleaned_text = re.sub(r"[^\w\sáéíóúàèìòùäëïöüñç]", "", text)
    words = cleaned_text.split()
    
    if len(words) == 0:
        return {
            "error": "No valid words found in the file."
        }
    
    word_frequencies = Counter(words)
    organized_words = defaultdict(dict)
    
    for word in sorted(word_frequencies):
        first_letter = word[0]
        organized_words[first_letter][word] = word_frequencies[word]
    
    return dict(sorted(organized_words.items()))
```

## Ejercicio 2: Conversión de tipos de datos

El objetivo de este ejercicio es crear una función que reciba una lista con diferentes tipos de datos y los separe en dos grupos:

- Valores que se pueden convertir a `float`
- Valores que no se pueden convertir a `float`

La lista puede contener elementos simples, listas internas o tuplas internas.  
Por este motivo, primero se aplana la estructura para poder analizar todos los valores individualmente.

```python
def flatten(items):
    """
    Recursively flatten lists and tuples.

    Parameters:
    items (list or tuple): A nested structure containing values, lists or tuples.

    Returns:
    list: A flat list with all individual elements.
    """
    
    flattened_items = []
    
    for item in items:
        if isinstance(item, (list, tuple)):
            flattened_items.extend(flatten(item))
        else:
            flattened_items.append(item)
    
    return flattened_items


def convert_values_to_float(data):
    """
    Convert all possible values from a nested data structure into floats.

    Parameters:
    data (list): A list that may contain values, lists or tuples.

    Returns:
    tuple: A tuple with two lists:
           - values converted to float
           - values that could not be converted
    """
    
    if not isinstance(data, list):
        return {
            "error": "Input must be a list."
        }
    
    converted_values = []
    not_converted_values = []
    
    flat_data = flatten(data)
    
    for item in flat_data:
        try:
            converted_values.append(float(item))
        except (ValueError, TypeError):
            not_converted_values.append(item)
    
    return converted_values, not_converted_values
```

## Ejercicio 1: Generador de contraseñas

El objetivo de este ejercicio es crear una función que genere contraseñas aleatorias y más seguras.

La función permite configurar diferentes criterios:

- Longitud de la contraseña
- Inclusión de letras mayúsculas
- Inclusión de letras minúsculas
- Inclusión de números
- Inclusión de signos o caracteres especiales

La función debe asegurar que todos los criterios seleccionados se cumplan al menos una vez en la contraseña generada.

```python
def crear_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    """
    Generate a random password based on selected character groups.

    Parameters:
    longitud (int): Password length.
    majuscules (bool): Include uppercase letters.
    minuscules (bool): Include lowercase letters.
    numeros (bool): Include numbers.
    signes (bool): Include special characters.

    Returns:
    str or dict: Generated password, or an error message for invalid inputs.
    """
    
    if not isinstance(longitud, int):
        return {
            "error": "Password length must be an integer."
        }
    
    if longitud <= 0:
        return {
            "error": "Password length must be greater than zero."
        }
    
    if not all(isinstance(option, bool) for option in [majuscules, minuscules, numeros, signes]):
        return {
            "error": "Character group options must be boolean values."
        }
    
    character_groups = {}
    
    if majuscules:
        character_groups["uppercase"] = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
    
    if minuscules:
        character_groups["lowercase"] = list("abcdefghijklmnopqrstuvwxyz")
    
    if numeros:
        character_groups["numbers"] = list("0123456789")
    
    if signes:
        character_groups["symbols"] = list(",-$?¿!@#%&*")
    
    if len(character_groups) == 0:
        return {
            "error": "At least one character group must be selected."
        }
    
    if longitud < len(character_groups):
        return {
            "error": "Password length is too short for the selected criteria."
        }
    
    password_characters = []
    
    for group in character_groups.values():
        password_characters.append(np.random.choice(group))
    
    all_characters = []
    
    for group in character_groups.values():
        all_characters.extend(group)
    
    remaining_length = longitud - len(password_characters)
    
    if remaining_length > 0:
        password_characters.extend(
            np.random.choice(all_characters, size=remaining_length)
        )
    
    np.random.shuffle(password_characters)
    
    password = "".join(password_characters)
    
    return password
```

### Comprobación de los criterios de la contraseña

Para comprobar que la contraseña generada cumple los criterios seleccionados, se define una función auxiliar que verifica si contiene mayúsculas, minúsculas, números y signos.

```python
def check_password_criteria(password):
    """
    Check which character groups are present in a password.

    Parameters:
    password (str): Password to check.

    Returns:
    dict: Boolean values indicating whether each character group is present.
    """
    
    return {
        "has_uppercase": any(char.isupper() for char in password),
        "has_lowercase": any(char.islower() for char in password),
        "has_number": any(char.isdigit() for char in password),
        "has_symbol": any(char in ",-$?¿!@#%&*" for char in password)
    }
```

## Ejercicio 2: Procesamiento simple de datos de fútbol

El objetivo de este ejercicio es procesar un archivo `.txt` que contiene resultados históricos de partidos de fútbol catalán.

A partir de este archivo, se deben obtener los siguientes resultados:

- El número total de goles marcados por cada equipo
- El equipo más goleador
- El equipo que más goles ha recibido
- La clasificación global, asignando:
  - 3 puntos por victoria
  - 1 punto por empate
  - 0 puntos por derrota

El archivo `historic partits.txt` se encuentra en la misma carpeta que el notebook, por lo que se utiliza una ruta relativa.

```python
def process_football_results(file_path):
    """
    Process a TXT file with football match results and calculate team statistics.

    Parameters:
    file_path (str): Path to the TXT file.

    Returns:
    dict: A dictionary containing:
          - total goals by team
          - highest scoring team
          - team with most goals conceded
          - global classification as a pandas DataFrame
    """
    
    if not isinstance(file_path, str):
        return {
            "error": "File path must be a string."
        }
    
    try:
        with open(file_path, "r", encoding="utf-8") as file:
            lines = file.readlines()
    except FileNotFoundError:
        return {
            "error": "File not found. Please check the file path."
        }
    except Exception as error:
        return {
            "error": f"An unexpected error occurred: {error}"
        }
    
    if len(lines) == 0:
        return {
            "error": "The file is empty."
        }
    
    team_stats = defaultdict(lambda: {
        "played": 0,
        "wins": 0,
        "draws": 0,
        "losses": 0,
        "goals_for": 0,
        "goals_against": 0,
        "points": 0
    })
    
    for line in lines:
        line = line.strip()
        
        if line == "":
            continue
        
        parts = line.split("\t")
        
        if len(parts) != 3:
            return {
                "error": f"Invalid line format: {line}"
            }
        
        home_team = parts[0]
        score = parts[1]
        away_team = parts[2]
        
        try:
            home_goals, away_goals = score.split("-")
            home_goals = int(home_goals)
            away_goals = int(away_goals)
        except ValueError:
            return {
                "error": f"Invalid score format: {score}"
            }
        
        team_stats[home_team]["played"] += 1
        team_stats[away_team]["played"] += 1
        
        team_stats[home_team]["goals_for"] += home_goals
        team_stats[home_team]["goals_against"] += away_goals
        
        team_stats[away_team]["goals_for"] += away_goals
        team_stats[away_team]["goals_against"] += home_goals
        
        if home_goals > away_goals:
            team_stats[home_team]["wins"] += 1
            team_stats[away_team]["losses"] += 1
            team_stats[home_team]["points"] += 3
            
        elif home_goals < away_goals:
            team_stats[away_team]["wins"] += 1
            team_stats[home_team]["losses"] += 1
            team_stats[away_team]["points"] += 3
            
        else:
            team_stats[home_team]["draws"] += 1
            team_stats[away_team]["draws"] += 1
            team_stats[home_team]["points"] += 1
            team_stats[away_team]["points"] += 1
    
    classification = pd.DataFrame.from_dict(team_stats, orient="index")
    classification = classification.reset_index().rename(columns={"index": "team"})
    
    classification = classification.sort_values(
        by=["points", "goals_for"],
        ascending=[False, False]
    ).reset_index(drop=True)
    
    total_goals_by_team = dict(
        zip(classification["team"], classification["goals_for"])
    )
    
    highest_scoring_team = classification.loc[
        classification["goals_for"].idxmax(),
        "team"
    ]
    
    most_conceded_team = classification.loc[
        classification["goals_against"].idxmax(),
        "team"
    ]
    
    return {
        "total_goals_by_team": total_goals_by_team,
        "highest_scoring_team": highest_scoring_team,
        "most_conceded_team": most_conceded_team,
        "classification": classification
    }
```
